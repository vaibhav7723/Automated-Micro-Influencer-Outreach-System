```python
import discovery
import filtering
import enrichment
import personalization
import sending

raw = discovery.run_discovery()

classified = filtering.filter_influencers(raw)

passed = []

for record in classified:
    if record["status"] == "PASS":
        passed.append(record)

enriched = enrichment.enrich_influencers(passed)

personalized = personalization.personalize_influencers(enriched)

results = sending.run_sending_layer(personalized)

results
```
